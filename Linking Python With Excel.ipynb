### Linking Python With Excel

An Excel table is read using the xlwings library to calculate the profit after tax of ten retail stores, and the outputs are directly updated on Excel by integrating Python with Excel using a VBA Button.

```python
#Import Libraries
import pandas as pd
import numpy as np
import xlwings as xw
```

```python
#Reading and Updating Excel Data
@xw.sub
def net_profit_calculator():
    wb=xw.Book.caller()
    sheet=wb.sheets["Input"]
    
    df_retail=sheet.range("A1").options(pd.DataFrame,expand="table").value
    df_retail['Gross Profit'] = df_retail['Total Revenue (Last FY)'] - df_retail['Total Expenses (Last FY)']
    df_retail['Tax Percentage']=0.25
    df_retail['Tax Amount']=df_retail['Gross Profit']*df_retail['Tax Percentage']
    df_retail['Net Profit']=df_retail['Gross Profit']-df_retail['Tax Amount']

    wb.sheets["Output"].range("A1").value=df_retail   
```
